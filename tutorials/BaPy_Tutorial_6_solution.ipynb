### Tutorial on Data Visualization with Matplotlib

## **Part 1: Introduction to Matplotlib**

**Matplotlib** is the most widely used plotting library in Python. It allows you to create a wide variety of static, animated, and interactive visualizations. Whether you are exploring data, presenting results, or preparing figures for a research paper, Matplotlib is an essential tool in your toolkit.

Why does visualization matter?
- Humans are visual creatures -- a well-designed plot can reveal patterns that are invisible in raw numbers
- In cognitive science, plots are crucial for communicating experimental results
- Visualization helps you catch errors in your data early

We always import Matplotlib's `pyplot` module with the conventional alias `plt`:

````python
import matplotlib.pyplot as plt
import numpy as np
````

Let's create our very first plot. `plt.plot()` takes a list of values and draws a line through them. In Jupyter notebooks, the plot is displayed automatically at the end of a cell. In regular Python scripts, you would need to call `plt.show()` to display the plot.

````python
# Our first plot -- just a list of numbers
plt.plot([1, 4, 2, 7, 5, 3, 6])
plt.show()
````

When you pass a single list to `plt.plot()`, Matplotlib uses the list index (0, 1, 2, ...) as the x-axis and the values as the y-axis. You can also call `plt.show()` explicitly, which is good practice and necessary outside of Jupyter.

## **Part 2: Line Plots**

Usually we want to plot y-values against specific x-values. We can pass both to `plt.plot(x, y)`.

````python
x = [0, 1, 2, 3, 4, 5]
y = [0, 1, 4, 9, 16, 25]

plt.plot(x, y)
plt.show()
````

**Customizing your plot**

Matplotlib gives you full control over how your plots look. Here are some common customizations:

- **Color**: Use the `color` parameter (e.g., `'red'`, `'blue'`, `'green'`, or hex codes like `'#FF5733'`)
- **Line style**: Use the `linestyle` parameter (e.g., `'-'` solid, `'--'` dashed, `':'` dotted, `'-.'` dash-dot)
- **Markers**: Use the `marker` parameter (e.g., `'o'` circles, `'s'` squares, `'^'` triangles, `'x'` crosses)
- **Line width**: Use the `linewidth` parameter

You can also add **labels** and a **title** to make your plot informative:

````python
x = [0, 1, 2, 3, 4, 5]
y = [0, 1, 4, 9, 16, 25]

plt.plot(x, y, color='red', linestyle='--', marker='o', linewidth=2)
plt.xlabel('x')
plt.ylabel('x squared')
plt.title('A Simple Quadratic Function')
plt.show()
````

## **Part 3: Scatter Plots**

Scatter plots are used to display individual data points. They are useful for showing the relationship between two variables. Use `plt.scatter(x, y)` to create one.

````python
# Basic scatter plot
x = [1, 2, 3, 4, 5, 6, 7]
y = [2, 3, 5, 4, 6, 8, 7]

plt.scatter(x, y)
plt.xlabel('x')
plt.ylabel('y')
plt.title('A Simple Scatter Plot')
plt.show()
````

You can customize scatter plots with:
- `s` -- marker size
- `c` or `color` -- marker color
- `marker` -- marker shape (e.g., `'o'`, `'s'`, `'^'`, `'D'`)
- `alpha` -- transparency (0 = fully transparent, 1 = fully opaque)
- `edgecolors` -- color of the marker border

````python
# Customized scatter plot
x = np.random.randn(50)
y = np.random.randn(50)

plt.scatter(x, y, s=100, c='coral', marker='D', alpha=0.7, edgecolors='black')
plt.xlabel('x')
plt.ylabel('y')
plt.title('Customized Scatter Plot')
plt.show()
````

## **Part 5: Histograms**

Histograms show the distribution of a dataset. They group data into **bins** and show how many data points fall into each bin. This is extremely useful in cognitive science for looking at reaction time distributions, neural firing rates, or any continuous measurement.

Use `plt.hist(data)` to create a histogram.

````python
# Basic histogram
data = np.random.randn(1000)  # 1000 samples from a standard normal distribution

plt.hist(data)
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.title('Histogram of Random Data')
plt.show()
````

The `bins` parameter controls how many bins the data is divided into. More bins give finer resolution, fewer bins give a smoother overview.

````python
# Effect of different bin counts
data = np.random.randn(1000)

plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
plt.hist(data, bins=10, color='skyblue', edgecolor='black')
plt.title('10 bins')

plt.subplot(1, 3, 2)
plt.hist(data, bins=30, color='salmon', edgecolor='black')
plt.title('30 bins')

plt.subplot(1, 3, 3)
plt.hist(data, bins=100, color='lightgreen', edgecolor='black')
plt.title('100 bins')

plt.tight_layout()
plt.show()
````

## **Part 6: Subplots**

Often you want to show multiple plots side by side. Matplotlib provides `plt.subplots()` to create a figure with a grid of axes (subplots).

The syntax is:
```python
fig, axes = plt.subplots(nrows, ncols, figsize=(width, height))
```

- `fig` is the overall figure object
- `axes` is an array of individual axes (subplots)
- For a single row or column, `axes` is a 1D array: access with `axes[i]`
- For multiple rows and columns, `axes` is a 2D array: access with `axes[i, j]`

````python
# Creating a 1x2 grid of subplots
fig, axes = plt.subplots(1, 2, figsize=(10, 4))

# Left plot
axes[0].plot([1, 2, 3, 4], [1, 4, 2, 3], color='blue', marker='o')
axes[0].set_title('Line Plot')
axes[0].set_xlabel('x')
axes[0].set_ylabel('y')

# Right plot
axes[1].bar(['A', 'B', 'C', 'D'], [3, 7, 2, 5], color='orange')
axes[1].set_title('Bar Chart')
axes[1].set_xlabel('Category')
axes[1].set_ylabel('Value')

plt.tight_layout()
plt.show()
````

**Important:** When working with subplots, you use methods on individual axes objects (e.g., `axes[0].plot()`, `axes[0].set_title()`) instead of the global `plt.plot()`, `plt.title()` functions. Notice the naming: `set_title()` instead of `title()`, `set_xlabel()` instead of `xlabel()`, etc.

````python
# Creating a 2x2 grid of subplots
fig, axes = plt.subplots(2, 2, figsize=(10, 8))

# Top-left: line plot
x = np.linspace(0, 10, 50)
axes[0, 0].plot(x, np.sin(x), color='blue')
axes[0, 0].set_title('Sine Wave')

# Top-right: scatter plot
axes[0, 1].scatter(np.random.randn(50), np.random.randn(50), color='red', alpha=0.5)
axes[0, 1].set_title('Random Scatter')

# Bottom-left: bar chart
axes[1, 0].bar(['A', 'B', 'C'], [5, 3, 7], color='green')
axes[1, 0].set_title('Bar Chart')

# Bottom-right: histogram
axes[1, 1].hist(np.random.randn(500), bins=20, color='purple', edgecolor='black')
axes[1, 1].set_title('Histogram')

plt.tight_layout()
plt.show()
````

## **Part 7: Customization**

To create professional-looking figures, you will often want to add legends, grids, set the figure size, and save your figures to files.

**Legends** identify different data series in a plot. Pass a `label` to each plot call and then call `plt.legend()`.

````python
x = np.linspace(0, 2 * np.pi, 100)

plt.plot(x, np.sin(x), label='sin(x)', color='blue')
plt.plot(x, np.cos(x), label='cos(x)', color='red')
plt.legend()  # Displays the legend using the labels
plt.show()
````

**Grids** help readers read values from your plot more precisely. Use `plt.grid(True)`.

````python
x = np.linspace(0, 10, 50)
plt.plot(x, x ** 2, color='green')
plt.grid(True)
plt.title('Plot with Grid')
plt.show()
````

**Figure size** can be controlled using `plt.figure(figsize=(width, height))` before your plot calls. The size is in inches.

````python
plt.figure(figsize=(10, 3))
plt.plot([1, 2, 3, 4, 5], [2, 4, 1, 5, 3])
plt.title('A Wide Figure')
plt.show()
````

**Saving figures** to files is done with `plt.savefig('filename.png')`. You can save in various formats: PNG, PDF, SVG, etc. Call `savefig()` *before* `plt.show()`, because `show()` clears the figure.

Use `dpi=300` for high-resolution figures suitable for publications, and `bbox_inches='tight'` to remove excess whitespace.

````python
plt.figure(figsize=(6, 4))
plt.plot([1, 2, 3], [1, 4, 9], 'ro-')
plt.title('Saved Figure Example')
plt.savefig('example_plot.png', dpi=300, bbox_inches='tight')
plt.show()
print("Figure saved as 'example_plot.png'")
````

## **Cognitive Science Application: Visualizing Experimental Results**

In cognitive science, effective data visualization is crucial for understanding experimental results. In this section, we will create the kinds of plots you would typically find in a cognitive science research paper.

We will visualize data from a simulated **Flanker Task** - a classic attention experiment where participants must respond to a central arrow while ignoring surrounding (flanking) arrows that are either pointing in the same direction (congruent: >>>>>) or the opposite direction (incongruent: >><>>).

**Exercise 8.1**: Generate synthetic Flanker task data.

Run the cell below to generate simulated reaction time and accuracy data for a Flanker task experiment. This will create the data we will visualize in the following exercises.

````python
# Set random seed for reproducibility
np.random.seed(42)

# Number of trials per condition
n_trials = 100

# Generate reaction time data (in milliseconds)
# Congruent condition: faster responses (mean=420ms, std=50ms)
rt_congruent = np.random.normal(loc=420, scale=50, size=n_trials)
# Incongruent condition: slower responses (mean=480ms, std=65ms)
rt_incongruent = np.random.normal(loc=480, scale=65, size=n_trials)

# Remove any negative reaction times (not realistic)
rt_congruent = np.clip(rt_congruent, 200, None)
rt_incongruent = np.clip(rt_incongruent, 200, None)

# Generate accuracy data
# Congruent: 97% correct
accuracy_congruent = np.random.choice([1, 0], size=n_trials, p=[0.97, 0.03])
# Incongruent: 89% correct
accuracy_incongruent = np.random.choice([1, 0], size=n_trials, p=[0.89, 0.11])

# Trial numbers (for time-course analysis)
trial_numbers = np.arange(1, n_trials + 1)

# Print summary statistics
print("=" * 50)
print("Flanker Task - Simulated Data Summary")
print("=" * 50)
print(f"Congruent RT:   Mean = {np.mean(rt_congruent):.1f} ms, SD = {np.std(rt_congruent):.1f} ms")
print(f"Incongruent RT: Mean = {np.mean(rt_incongruent):.1f} ms, SD = {np.std(rt_incongruent):.1f} ms")
print(f"Congruent Accuracy:   {np.mean(accuracy_congruent) * 100:.1f}%")
print(f"Incongruent Accuracy: {np.mean(accuracy_incongruent) * 100:.1f}%")
print("=" * 50)
````

**Exercise 8.2**: Create an RT distribution plot.

Create overlapping histograms showing the reaction time distributions for both conditions:
- Plot `rt_congruent` and `rt_incongruent` as histograms on the same axes
- Use `alpha=0.6` for transparency so both distributions are visible
- Use different colors for each condition (e.g., `'steelblue'` and `'coral'`)
- Use 20 bins
- Add a legend with labels `'Congruent'` and `'Incongruent'`
- Add x-label: `"Reaction Time (ms)"`
- Add y-label: `"Frequency"`
- Add title: `"RT Distribution by Condition"`

````python
# Exercise 8.2: RT distribution plot
plt.figure(figsize=(8, 5))
plt.hist(rt_congruent, bins=20, alpha=0.6, color='steelblue', label='Congruent', edgecolor='black')
plt.hist(rt_incongruent, bins=20, alpha=0.6, color='coral', label='Incongruent', edgecolor='black')
plt.xlabel('Reaction Time (ms)')
plt.ylabel('Frequency')
plt.title('RT Distribution by Condition')
plt.legend()
plt.show()
````

**Exercise 8.3**: Create a bar chart with error bars showing mean RT per condition.

- Calculate the mean RT for each condition using `np.mean()`
- Calculate the standard error of the mean (SEM) for each condition: `SEM = np.std(data) / np.sqrt(len(data))`
- Create a bar chart with two bars (one per condition) using `plt.bar()`
- Add error bars using the `yerr` parameter in `plt.bar()`
- Use `capsize=5` in `plt.bar()` to add caps to the error bars
- Add a y-label: `"Mean Reaction Time (ms)"`
- Add a title: `"Mean RT by Condition"`
- *Hint*: You can start the y-axis at a value above 0 using `plt.ylim(bottom, top)` to better show the difference between conditions

````python
# Exercise 8.3: Bar chart with error bars
mean_con = np.mean(rt_congruent)
mean_inc = np.mean(rt_incongruent)
sem_con = np.std(rt_congruent) / np.sqrt(len(rt_congruent))
sem_inc = np.std(rt_incongruent) / np.sqrt(len(rt_incongruent))

plt.figure(figsize=(6, 5))
plt.bar(['Congruent', 'Incongruent'], [mean_con, mean_inc], 
        yerr=[sem_con, sem_inc], capsize=5,
        color=['steelblue', 'coral'], edgecolor='black')
plt.ylabel('Mean Reaction Time (ms)')
plt.title('Mean RT by Condition')
plt.ylim(350, 550)
plt.show()
````

**Exercise 8.4**: Create a combined figure (the full picture).

Create a 2x2 subplot figure that brings everything together:

- **Top-left**: RT histograms for both conditions (like Exercise 8.2)
- **Top-right**: Bar chart of mean RTs with error bars (like Exercise 8.3)
- **Bottom-left**: Bar chart of accuracy (%) per condition
- **Bottom-right**: Scatter plot of RT vs. trial number for both conditions (to check for learning or fatigue effects)

Requirements:
- Use `fig, axes = plt.subplots(2, 2, figsize=(12, 10))`
- Add a main title with `fig.suptitle("Flanker Task Results", fontsize=16, fontweight='bold')`
- Give each subplot its own title, axis labels, and legend where appropriate
- Use `plt.tight_layout()` to prevent overlapping
- Save the figure as `"flanker_results.png"` with `dpi=300`

````python
# Exercise 8.4: Combined figure
fig, axes = plt.subplots(2, 2, figsize=(12, 10))
fig.suptitle("Flanker Task Results", fontsize=16, fontweight='bold')

# Top-left: RT histograms
axes[0, 0].hist(rt_congruent, bins=20, alpha=0.6, color='steelblue', label='Congruent', edgecolor='black')
axes[0, 0].hist(rt_incongruent, bins=20, alpha=0.6, color='coral', label='Incongruent', edgecolor='black')
axes[0, 0].set_xlabel('Reaction Time (ms)')
axes[0, 0].set_ylabel('Frequency')
axes[0, 0].set_title('RT Distribution')
axes[0, 0].legend()

# Top-right: Bar chart with error bars
mean_con = np.mean(rt_congruent)
mean_inc = np.mean(rt_incongruent)
sem_con = np.std(rt_congruent) / np.sqrt(len(rt_congruent))
sem_inc = np.std(rt_incongruent) / np.sqrt(len(rt_incongruent))
axes[0, 1].bar(['Congruent', 'Incongruent'], [mean_con, mean_inc],
               yerr=[sem_con, sem_inc], capsize=5,
               color=['steelblue', 'coral'], edgecolor='black')
axes[0, 1].set_ylabel('Mean RT (ms)')
axes[0, 1].set_title('Mean RT by Condition')
axes[0, 1].set_ylim(350, 550)

# Bottom-left: Accuracy bar chart
acc_con = np.mean(accuracy_congruent) * 100
acc_inc = np.mean(accuracy_incongruent) * 100
axes[1, 0].bar(['Congruent', 'Incongruent'], [acc_con, acc_inc],
               color=['steelblue', 'coral'], edgecolor='black')
axes[1, 0].set_ylabel('Accuracy (%)')
axes[1, 0].set_title('Accuracy by Condition')
axes[1, 0].set_ylim(70, 105)

# Bottom-right: RT vs trial number
axes[1, 1].scatter(trial_numbers, rt_congruent, alpha=0.5, s=15, color='steelblue', label='Congruent')
axes[1, 1].scatter(trial_numbers, rt_incongruent, alpha=0.5, s=15, color='coral', label='Incongruent')
axes[1, 1].set_xlabel('Trial Number')
axes[1, 1].set_ylabel('RT (ms)')
axes[1, 1].set_title('RT Over Time')
axes[1, 1].legend()

plt.tight_layout()
plt.savefig('flanker_results.png', dpi=300, bbox_inches='tight')
plt.show()
````
