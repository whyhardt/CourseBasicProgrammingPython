#### **Variable Scope**

The lifetime of a variable refers to the period of time during which the variable exists in memory. Variable scope refers to the visibility and accessibility of variables in different parts of a program. 

 **Note:** Understanding variable scope is important for writing efficient and bug-free code, as it helps prevent unintended side effects and allows for better organization and management of variables within a program.

````python
def my_func():
    x = 10
    def inner_func():
        nonlocal x
        x = 5
    inner_func()
    print("Inside function, x =", x)

x = 20
my_func()
print("Outside function, x =", x)
````

**Note 1:** In this example, we have a variable `x` defined outside the function, another variable `x` defined inside the function `my_func`, and another function `inner_func` inside the first function (also called: nested function) that modifies the value of `x`. By using the `nonlocal` keyword inside `inner_func`, we can access the variable `x` defined in my_func and modify its value.

**Note 2:** To define a global variable inside a function, you can use the `global` keyword followed by the name of the variable. This tells Python to use the global variable with that name, rather than creating a new local variable with the same name.

**Note 3:** If we did not use the `global` keyword inside `my_function`, the x variable would be treated as a local variable, and would not affect the value of the `global x` variable outside of the function.

````python
x = 10

def my_function():
    global x
    x = 20
    print("Inside the function, x is:", x)

my_function()
print("Outside the function, x is:", x)
````

**Recursion**

One function can call another. A function can also call itself. It may not be obvious why that is a good thing, but it turns out to be one of the most magical things a program can do. A function that calls itself is called **recursive**. The process is called **recursion**. Recursion is a programming technique that involves a function calling itself within its own definition. Recursive functions can be used to solve problems that involve repeated sub-tasks or sub-problems, such as traversing a tree or searching for a path in a graph.

````python
def countdown(n):
  if n <= 0:
    print ('Stop!')
  else:
    print(n)
    countdown(n - 1)

countdown(3)
````

Implement a recursive function to calculate the sum of digits of a given number.

````python
def sum_of_digits(n):
    if n <= 0:
        return 0
    return n + sum_of_digits(n - 1)

print(sum_of_digits(5))
````

Implement a recursive function to calculate the factorial of a given number

````python
def factorial(n):
    if n <= 1:
        return 1
    return n * factorial(n - 1)

print(factorial(5))
````

#### **Docstrings**

In Python, a docstring is a way to document what a function does, what parameters it takes, what it returns, and any other relevant information. It's like a little note that helps you and others understand how to use the function correctly.

Why Are Docstrings Useful?

- Documentation: Docstrings serve as documentation for your code. They help you and others understand what a function does without having to read through the entire code.
- Readability: Well-written docstrings make your code more readable. They provide context and clarity, making it easier for others (or yourself in the future) to understand and use your code.
- Interactive Help: When you're working in an interactive environment like Jupyter Notebook or IPython, you can use docstrings to quickly get help about a function. Just type the function's name followed by a question mark (function_name?) to see its docstring.

Docstrings always have the same structure which is:
1. Explanation of what that function does
2. List of input arguments and their types started with `Args:`
3. List of return values with only their types and a short description started with `Returns:` 

Here's an example on how docstrings might look like. 

````python
def add(a, b):
    """Add two numbers together.
    
    Args:
        a (int): The first number.
        b (int): The second number.
    
    Returns:
        int: The sum of a and b.
    """
    return a + b
````

## **Cognitive Science Application: Stroop Task Simulator**

Remember the Stroop experiment from Tutorial 3? You built a working simulation using loops and conditionals — but the code was one long block that would be painful to reuse or modify.

Now let's **refactor** that same experiment using **functions**. You'll see how breaking code into named, reusable pieces makes it:
- **Easier to read** — each function does one thing
- **Easier to reuse** — call it as many times as you want with different parameters
- **Easier to test** — you can check each piece independently

We'll build the same simulator with four functions:
1. `create_trial()` — generate a single congruent or incongruent trial
2. `simulate_response()` — simulate a participant's reaction
3. `run_experiment()` — run many trials and collect results
4. `analyze_results()` — compute summary statistics

Compare this to your Tutorial 3 code as you go — the logic is the same, but the structure is completely different.

**Step 1: Setting up the experiment**

First, we need to define what colors and words our Stroop task will use. We also need a function that creates a single trial -- either congruent (word matches ink color) or incongruent (word and ink color differ).

Write a function `create_trial(colors, is_congruent)` that:
- Takes a list of color names and a boolean `is_congruent`
- Picks a random color word using `random.choice()`
- If congruent: the ink color is the same as the word
- If incongruent: the ink color is a *different* random color from the list
- Returns both the `word` and the `ink_color`

*Hint: For incongruent trials, you could keep picking a random color until it differs from the word, or remove the word from the list first.*

````python
import random

colors = ["red", "green", "blue", "yellow"]

def create_trial(colors, is_congruent):
    """Create a single Stroop trial.
    
    Args:
        colors (list): List of color name strings.
        is_congruent (bool): Whether the trial should be congruent.
    
    Returns:
        tuple: (word, ink_color)
    """
    # Pick a random color word
    word = random.choice(colors)
    
    if is_congruent:
        ink_color = word
    else:
        # Pick a different color for the ink
        other_colors = [c for c in colors if c != word]
        ink_color = random.choice(other_colors)
    
    return word, ink_color

# Test your function
print("Congruent trial:", create_trial(colors, True))
print("Incongruent trial:", create_trial(colors, False))
````

**Step 2: Simulating a participant's response**

In a real experiment, participants take time to respond. In the Stroop task:
- **Congruent trials** are faster (word and color match, no conflict)
- **Incongruent trials** are slower (the brain has to resolve the conflict between reading the word and naming the color)

Write a function `simulate_response(word, ink_color)` that:
- Determines if the trial is congruent or incongruent
- Simulates a response time (in milliseconds):
  - Congruent: base time of **500ms** + random variation between 0 and 100ms
  - Incongruent: base time of **650ms** + random variation between 0 and 150ms
- Simulates accuracy: the participant sometimes makes mistakes
  - Congruent: **95%** chance of being correct
  - Incongruent: **85%** chance of being correct
- Returns the `response_time` and whether the response was `correct` (True/False)

*Hint: Use `random.randint(a, b)` for the random variation and `random.random()` (returns a float between 0 and 1) to simulate accuracy.*

````python
def simulate_response(word, ink_color):
    """Simulate a participant's response to a Stroop trial.
    
    Args:
        word (str): The written color word.
        ink_color (str): The ink color the word is displayed in.
    
    Returns:
        tuple: (response_time, correct)
    """
    if word == ink_color:
        # Congruent: fast (500ms base) + small random variation (0-100ms)
        response_time = 500 + random.randint(0, 100)
        # 95% chance of being correct
        correct = random.random() < 0.95
    else:
        # Incongruent: slower (650ms base) + larger random variation (0-150ms)
        response_time = 650 + random.randint(0, 150)
        # 85% chance of being correct
        correct = random.random() < 0.85
    
    return response_time, correct

# Test your function
word, ink = create_trial(colors, True)
rt, correct = simulate_response(word, ink)
print(f"Congruent - Word: {word}, Ink: {ink}, RT: {rt}ms, Correct: {correct}")

word, ink = create_trial(colors, False)
rt, correct = simulate_response(word, ink)
print(f"Incongruent - Word: {word}, Ink: {ink}, RT: {rt}ms, Correct: {correct}")
````

**Step 3: Running the full experiment**

Now let's put it all together! Write a function `run_experiment(colors, n_trials)` that:
- Runs `n_trials` trials, with half being congruent and half incongruent (randomized)
- For each trial, creates a trial and simulates a response
- Collects all results into four separate lists:
  - `congruent_rts` - reaction times for congruent trials
  - `incongruent_rts` - reaction times for incongruent trials
  - `congruent_correct` - accuracy (True/False) for congruent trials
  - `incongruent_correct` - accuracy (True/False) for incongruent trials
- Returns all four lists

*Hint: You can create a list of `True` and `False` values for congruency, then use `random.shuffle()` to randomize trial order.*

````python
def run_experiment(colors, n_trials):
    """Run a full Stroop experiment.
    
    Args:
        colors (list): List of color name strings.
        n_trials (int): Total number of trials (split evenly between conditions).
    
    Returns:
        tuple: (congruent_rts, incongruent_rts, congruent_correct, incongruent_correct)
    """
    # Prepare empty lists to collect results
    congruent_rts = []
    incongruent_rts = []
    congruent_correct = []
    incongruent_correct = []
    
    # Create a list of trial types: half True (congruent), half False (incongruent)
    trial_types = [True] * (n_trials // 2) + [False] * (n_trials // 2)
    random.shuffle(trial_types)
    
    # Run each trial
    for is_congruent in trial_types:
        # 1. Create the trial
        word, ink_color = create_trial(colors, is_congruent)
        
        # 2. Simulate the response
        rt, correct = simulate_response(word, ink_color)
        
        # 3. Store results in the appropriate lists
        if is_congruent:
            congruent_rts.append(rt)
            congruent_correct.append(correct)
        else:
            incongruent_rts.append(rt)
            incongruent_correct.append(correct)
    
    return congruent_rts, incongruent_rts, congruent_correct, incongruent_correct

# Run experiment with 100 trials
congruent_rts, incongruent_rts, congruent_correct, incongruent_correct = run_experiment(colors, 100)
````

**Step 4: Analyzing the results**

Now write a function `analyze_results(rts, correct_list, condition_name)` that:
- Takes a list of reaction times, a list of correct/incorrect booleans, and a condition name string
- Calculates and prints:
  - The **mean reaction time** (sum of RTs divided by number of trials)
  - The **accuracy** (percentage of correct responses)
- Returns the mean RT and accuracy

Then call it for both conditions and compare!

````python
def analyze_results(rts, correct_list, condition_name):
    """Analyze results for one condition of the Stroop experiment.
    
    Args:
        rts (list): List of reaction times in ms.
        correct_list (list): List of booleans indicating correctness.
        condition_name (str): Name of the condition (e.g. "Congruent").
    
    Returns:
        tuple: (mean_rt, accuracy_percentage)
    """
    # Calculate mean RT: sum of all RTs divided by number of trials
    mean_rt = sum(rts) / len(rts)
    
    # Calculate accuracy: count True values, divide by total, multiply by 100
    accuracy = sum(correct_list) / len(correct_list) * 100
    
    print(f"{condition_name}: Mean RT = {mean_rt:.1f} ms, Accuracy = {accuracy:.1f}%")
    return mean_rt, accuracy

# Analyze both conditions
print("=" * 40)
print("STROOP EXPERIMENT RESULTS")
print("=" * 40)
con_rt, con_acc = analyze_results(congruent_rts, congruent_correct, "Congruent")
inc_rt, inc_acc = analyze_results(incongruent_rts, incongruent_correct, "Incongruent")
print("=" * 40)
print(f"\nStroop Effect (RT difference): {inc_rt - con_rt:.1f} ms")
print(f"Accuracy difference: {con_acc - inc_acc:.1f}%")
````

**Bonus: Simulate multiple participants**

In real experiments, we test many participants, not just one. Write a function `run_multi_participant(colors, n_participants, n_trials_per_participant)` that:
- Runs the experiment for each participant
- Collects each participant's mean congruent RT and mean incongruent RT
- Prints the **grand average** across all participants for both conditions
- Prints how many participants showed a Stroop effect (incongruent RT > congruent RT)

This is similar to how cognitive scientists analyze group-level data!

````python
def run_multi_participant(colors, n_participants, n_trials_per_participant):
    """Run the Stroop experiment for multiple participants and report group results."""
    
    # Collect each participant's mean RTs
    all_congruent_means = []
    all_incongruent_means = []
    
    for participant in range(n_participants):
        # Run the experiment for this participant
        con_rts, inc_rts, con_cor, inc_cor = run_experiment(colors, n_trials_per_participant)
        
        # Calculate this participant's mean RT for each condition and store it
        all_congruent_means.append(sum(con_rts) / len(con_rts))
        all_incongruent_means.append(sum(inc_rts) / len(inc_rts))
    
    # Grand averages across all participants
    grand_con = sum(all_congruent_means) / len(all_congruent_means)
    grand_inc = sum(all_incongruent_means) / len(all_incongruent_means)
    
    print(f"Grand average congruent RT:   {grand_con:.1f} ms")
    print(f"Grand average incongruent RT: {grand_inc:.1f} ms")
    print(f"Grand Stroop effect:          {grand_inc - grand_con:.1f} ms")
    
    # Count how many participants showed a Stroop effect
    n_stroop = 0
    for i in range(n_participants):
        if all_incongruent_means[i] > all_congruent_means[i]:
            n_stroop += 1
    print(f"\nParticipants showing Stroop effect: {n_stroop}/{n_participants}")

# Simulate 20 participants, each doing 50 trials
run_multi_participant(colors, 20, 50)
````
